# Step 1: Import Python Libraries

Python libraries are imported to help perform:
- Data analysis
- Data visualisation
- Machine learning
- Model evaluation

Each library has a specific purpose.


```python

# Import Pandas
# Pandas is used for data manipulation and working with tables/dataframes
import pandas as pd

# Import NumPy
# NumPy is used for numerical operations and mathematical calculations
import numpy as np

# Import Matplotlib
# Matplotlib is used for creating charts and visualisations
import matplotlib.pyplot as plt

# Import Seaborn
# Seaborn is used for advanced statistical visualisations
import seaborn as sns

# Import LabelEncoder
# LabelEncoder converts categorical text values into numerical values
from sklearn.preprocessing import LabelEncoder

# Import StandardScaler
# StandardScaler normalises numerical data so that features are on similar scales
from sklearn.preprocessing import StandardScaler

# Import train_test_split
# This splits the dataset into training and testing datasets
from sklearn.model_selection import train_test_split

# Import machine learning algorithms
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# Import evaluation metrics
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import roc_auc_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
from sklearn.metrics import roc_curve

# Ignore warnings to make notebook cleaner
import warnings
warnings.filterwarnings('ignore')
```


# Step 2: Load the Dataset

The dataset is loaded using Pandas.

Why do we load the dataset?
- To analyse patient information
- To prepare the data for machine learning
- To inspect the dataset structure

The dataset file must be uploaded into Databricks before running the notebook.


```python

# Load the dataset using Pandas
# Replace the file path if necessary when uploading to Databricks

df = pd.read_csv('Expanded_Patient_Readmission_Data_(1).csv')

# Display the first 5 rows
# head() helps us preview the dataset structure
df.head()
```


# Step 3: Understand the Dataset

Before building a machine learning model, we first inspect:
- Number of rows and columns
- Data types
- Missing values
- General dataset structure


```python

# Display the number of rows and columns
# shape returns (rows, columns)
print("Dataset Shape:")
print(df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns)

# Display data types
# info() helps identify numerical and categorical columns
print("\nDataset Information:")
df.info()
```


# Step 4: Check for Missing Values

Missing values can negatively affect machine learning models.

We check for missing values because:
- Models cannot handle missing values properly
- Missing data can reduce model accuracy
- We may need to clean or fill missing values


```python

# Check for missing values in each column
# isnull() identifies missing values
# sum() counts them

print("Missing Values Per Column:")
print(df.isnull().sum())
```


# Step 5: Check for Duplicate Records

Duplicate records can:
- Bias the machine learning model
- Create incorrect patterns
- Reduce model reliability


```python

# Check for duplicate rows
duplicates = df.duplicated().sum()

# Display number of duplicates
print("Number of Duplicate Rows:", duplicates)
```


# Step 6: Encode Categorical Variables

Machine learning models cannot work directly with text values.

Examples of categorical variables:
- Gender
- Admission Type
- Readmission

We use Label Encoding to convert text into numbers.


```python

# Create LabelEncoder object
encoder = LabelEncoder()

# Encode Gender column
# Female/Male become numerical values
df['Gender'] = encoder.fit_transform(df['Gender'])

# Encode Admission Type column
df['Admission Type'] = encoder.fit_transform(df['Admission Type'])

# Encode Readmission target variable
# No = 0 and Yes = 1
df['Readmission'] = encoder.fit_transform(df['Readmission'])

# Display first 5 rows after encoding
df.head()
```


# Step 7: Define Features and Target Variable

Features (X):
- Independent variables used to make predictions

Target Variable (y):
- Variable we want to predict

In this project:
- X = patient information
- y = readmission status


```python

# Create feature dataset X
# Drop Patient ID because it does not help prediction
# Drop Readmission because it is the target variable

X = df.drop(['Patient ID', 'Readmission'], axis=1)

# Create target variable y
y = df['Readmission']

# Display feature columns
print("Feature Columns:")
print(X.columns)

# Display target variable
print("\nTarget Variable:")
print(y.head())
```


# Step 8: Normalise the Data

Machine learning models perform better when numerical variables are on similar scales.

Why scaling is important:
- Prevents large numbers from dominating smaller numbers
- Improves model performance
- Helps optimisation algorithms converge faster

We use StandardScaler:
- Mean becomes 0
- Standard deviation becomes 1


```python

# Create StandardScaler object
scaler = StandardScaler()

# Scale the feature dataset
X_scaled = scaler.fit_transform(X)

# Convert scaled data back into dataframe for readability
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)

# Display first 5 rows
X_scaled.head()
```


# Step 9: Exploratory Data Analysis (EDA)

EDA helps us:
- Understand data distributions
- Detect trends and relationships
- Identify important predictors

We use visualisations to better understand the dataset.


```python

# Display statistical summary
# describe() provides statistics like mean, minimum, maximum, and standard deviation

df.describe()
```

```python

# Create a count plot for Readmission

plt.figure(figsize=(6,4))

# countplot shows distribution of categorical values
sns.countplot(x='Readmission', data=df)

# Add title
plt.title('Distribution of Readmission')

# Show plot
plt.show()
```

```python

# Create histogram for Age distribution

plt.figure(figsize=(8,5))

# histplot shows distribution of numerical values
sns.histplot(df['Age'], bins=20, kde=True)

# Add title
plt.title('Age Distribution')

# Display chart
plt.show()
```

```python

# Create boxplot for Length of Stay vs Readmission

plt.figure(figsize=(8,5))

# boxplot helps identify outliers and distributions
sns.boxplot(x='Readmission', y='Length of Stay', data=df)

# Add title
plt.title('Length of Stay vs Readmission')

# Show chart
plt.show()
```

```python

# Create correlation heatmap

plt.figure(figsize=(10,8))

# corr() calculates correlations between variables
correlation_matrix = df.corr()

# heatmap visualises relationships between variables
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')

# Add title
plt.title('Correlation Heatmap')

# Show plot
plt.show()
```


# Step 10: Feature Selection

Feature selection helps identify the most important predictors.

Why feature selection is important:
- Improves model performance
- Reduces overfitting
- Removes unnecessary variables
- Makes the model faster

We use correlation analysis to determine important features.


```python

# Calculate correlation with target variable

feature_importance = df.corr()['Readmission'].sort_values(ascending=False)

# Display correlations
print(feature_importance)
```


## Feature Selection Discussion

Features with higher correlation values are considered more important predictors.

Examples:
- Previous Admissions
- Length of Stay
- Number of Diagnoses

These variables may strongly influence whether a patient is readmitted.

Features with very low correlations may contribute less to prediction accuracy.



# Step 11: Split Data into Training and Testing Sets

The dataset is divided into:
- Training Data = used to train the model
- Testing Data = used to evaluate the model

Why do we split data?
- To test model performance on unseen data
- To avoid overfitting


```python

# Split dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,          # Feature dataset
    y,                 # Target variable
    test_size=0.2,     # 20% testing data
    random_state=42    # Ensures reproducibility
)

# Display shapes
print("Training Features Shape:", X_train.shape)
print("Testing Features Shape:", X_test.shape)
```


# Step 12: Build Machine Learning Models

We will build three models:
1. Logistic Regression
2. Decision Tree
3. Random Forest

Using multiple models helps compare performance and choose the best one.


```python

# Create Logistic Regression model
log_model = LogisticRegression()

# Train the model using training data
log_model.fit(X_train, y_train)

# Make predictions using testing data
y_pred_log = log_model.predict(X_test)

# Display predictions
print(y_pred_log[:10])
```

```python

# Evaluate Logistic Regression model

print("Logistic Regression Accuracy:",
      accuracy_score(y_test, y_pred_log))

print("Precision:",
      precision_score(y_test, y_pred_log))

print("Recall:",
      recall_score(y_test, y_pred_log))

print("F1 Score:",
      f1_score(y_test, y_pred_log))

print("ROC-AUC:",
      roc_auc_score(y_test, y_pred_log))
```

```python

# Create Decision Tree model
tree_model = DecisionTreeClassifier(random_state=42)

# Train model
tree_model.fit(X_train, y_train)

# Make predictions
y_pred_tree = tree_model.predict(X_test)

# Evaluate model
print("Decision Tree Accuracy:",
      accuracy_score(y_test, y_pred_tree))

print("Precision:",
      precision_score(y_test, y_pred_tree))

print("Recall:",
      recall_score(y_test, y_pred_tree))

print("F1 Score:",
      f1_score(y_test, y_pred_tree))

print("ROC-AUC:",
      roc_auc_score(y_test, y_pred_tree))
```

```python

# Create Random Forest model
forest_model = RandomForestClassifier(
    n_estimators=100,   # Number of trees
    random_state=42
)

# Train Random Forest model
forest_model.fit(X_train, y_train)

# Make predictions
y_pred_forest = forest_model.predict(X_test)

# Evaluate Random Forest model
print("Random Forest Accuracy:",
      accuracy_score(y_test, y_pred_forest))

print("Precision:",
      precision_score(y_test, y_pred_forest))

print("Recall:",
      recall_score(y_test, y_pred_forest))

print("F1 Score:",
      f1_score(y_test, y_pred_forest))

print("ROC-AUC:",
      roc_auc_score(y_test, y_pred_forest))
```


# Step 13: Compare Model Performance

The best model is usually the one with:
- Higher accuracy
- Better precision and recall
- Higher ROC-AUC score

Random Forest often performs better because:
- It combines multiple decision trees
- Reduces overfitting
- Handles complex relationships well


```python

# Create comparison table

results = pd.DataFrame({
    'Model': ['Logistic Regression',
              'Decision Tree',
              'Random Forest'],

    'Accuracy': [
        accuracy_score(y_test, y_pred_log),
        accuracy_score(y_test, y_pred_tree),
        accuracy_score(y_test, y_pred_forest)
    ]
})

# Display results
results
```


# Step 14: Confusion Matrix

A confusion matrix helps us understand:
- Correct predictions
- Incorrect predictions
- False positives
- False negatives


```python

# Generate confusion matrix for Random Forest

cm = confusion_matrix(y_test, y_pred_forest)

# Create heatmap for confusion matrix

plt.figure(figsize=(6,4))

sns.heatmap(cm,
            annot=True,
            fmt='d',
            cmap='Blues')

# Add labels and title
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix - Random Forest')

# Show plot
plt.show()
```


# Step 15: Overfitting Discussion

## What is Overfitting?

Overfitting occurs when:
- A model memorises training data
- The model performs poorly on new data

Signs of overfitting:
- Very high training accuracy
- Low testing accuracy

## Strategies to Reduce Overfitting

1. Cross-validation
2. Feature selection
3. Using more data
4. Hyperparameter tuning
5. Pruning decision trees
6. Regularisation techniques

Random Forest helps reduce overfitting because:
- It averages multiple trees
- Reduces variance
